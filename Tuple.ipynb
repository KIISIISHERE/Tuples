```python
# 1. Create a tuple with different datatypes
mixed_tuple = (10, "Hello", 3.14, True)
print("Tuple with different datatypes:", mixed_tuple)
# 2. Access the second element of the tuple
int_tuple = (1, 2, 3, 4, 5)
print("Integer tuple:", int_tuple)
# 3. Create a new tuple by adding 9 to each element in the integer tuple
# Using a generator expression with tuple comprehension
updated_tuple = tuple(x + 9 for x in int_tuple)
print("Tuple after adding 9 to each element:", updated_tuple)
# 4. Count the occuurences of an element in the tuple
# For example, count how many times '12' appears in the updated tuple
element_to_count = 12
count = updated_tuple.count(element_to_count)
print(f"Occurrences {element_to_count}:", count)
# 5. Perform slicing on the tuple
# For example, take elements from index 1 to 3
sliced_tuple = updated_tuple[1:4]
print("Sliced tuple (index 1 to 3):", sliced_tuple)
```

```python
# Given tuple
t = (1, 2, 3, 3, 2, 1)
# Check if the tuple is the sane when reversed
if t == t[::-1]:
    print("The tuple is a palindrome.")
else:
    print("The tuple is not a palindrome.")
```

```python
weather = (1, 0, 0, 0, 1, 1, 0)
rainy = weather.count(1)
sunny = weather.count(0)
print("Rainy days:", rainy)
print("Sunny days:", sunny)
if rainy > sunny:
  print("Predicted weather: Rainy")
elif sunny > rainy:
  print("Predicted weather: Sunny")
else:
  print("Predicted weather: Uncertain")
```
